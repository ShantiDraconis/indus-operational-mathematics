# Indus Script Frequency Analysis

This notebook performs frequency analysis on the Indus script corpus to validate the operational model.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
```

## Load Data

```python
# Load datasets
sign_list = pd.read_csv('../data/sign_list.csv', comment='#')
frequencies = pd.read_csv('../data/frequencies.csv', comment='#')
bigrams = pd.read_csv('../data/bigrams.csv', comment='#')
site_dist = pd.read_csv('../data/site_distribution.csv', comment='#')
object_dist = pd.read_csv('../data/object_distribution.csv', comment='#')

print(f"Loaded {len(sign_list)} signs")
print(f"Loaded {len(frequencies)} frequency entries")
print(f"Loaded {len(bigrams)} bigram patterns")
```

## Sign Frequency Distribution

```python
# Plot frequency distribution
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.bar(range(len(frequencies)), frequencies['total_occurrences'])
plt.xlabel('Sign Rank')
plt.ylabel('Frequency')
plt.title('Sign Frequency Distribution (Zipf-like)')
plt.yscale('log')

plt.subplot(1, 2, 2)
cumulative = np.cumsum(frequencies['percentage'])
plt.plot(range(len(cumulative)), cumulative, linewidth=2)
plt.xlabel('Number of Signs')
plt.ylabel('Cumulative Percentage')
plt.title('Cumulative Distribution')
plt.axhline(y=50, color='r', linestyle='--', label='50%')
plt.axhline(y=80, color='g', linestyle='--', label='80%')
plt.legend()

plt.tight_layout()
plt.show()
```

## Positional Distribution

```python
# Analyze positional preferences
positions = sign_list[['initial_position', 'medial_position', 'terminal_position']].head(10)
positions.index = sign_list['sign_id'].head(10)

positions.plot(kind='bar', stacked=False, figsize=(12, 6))
plt.title('Positional Distribution of Top 10 Signs')
plt.xlabel('Sign')
plt.ylabel('Probability')
plt.xticks(rotation=45)
plt.legend(['Initial', 'Medial', 'Terminal'])
plt.tight_layout()
plt.show()
```

## Bigram Analysis

```python
# Most common bigrams
top_bigrams = bigrams.head(10)

plt.figure(figsize=(12, 6))
plt.barh(range(len(top_bigrams)), top_bigrams['count'])
plt.yticks(range(len(top_bigrams)), 
           [f"{row['sign1']}-{row['sign2']}" for _, row in top_bigrams.iterrows()])
plt.xlabel('Count')
plt.title('Top 10 Most Frequent Bigrams')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

print("\nOperational interpretations:")
for _, row in top_bigrams.iterrows():
    print(f"{row['sign1']}-{row['sign2']}: {row['interpretation']}")
```

## Site Distribution

```python
# Geographic distribution
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.bar(site_dist['site_name'], site_dist['total_inscriptions'])
plt.xticks(rotation=45, ha='right')
plt.ylabel('Number of Inscriptions')
plt.title('Inscriptions by Site')

plt.subplot(1, 2, 2)
object_types = site_dist[['seals', 'tablets', 'pottery', 'other']].sum()
plt.pie(object_types, labels=object_types.index, autopct='%1.1f%%')
plt.title('Distribution by Object Type')

plt.tight_layout()
plt.show()
```

## Statistical Summary

```python
print("Corpus Statistics:")
print(f"Total sites: {len(site_dist)}")
print(f"Total inscriptions: {site_dist['total_inscriptions'].sum()}")
print(f"Seals: {site_dist['seals'].sum()} ({site_dist['seals'].sum()/site_dist['total_inscriptions'].sum()*100:.1f}%)")
print(f"Tablets: {site_dist['tablets'].sum()} ({site_dist['tablets'].sum()/site_dist['total_inscriptions'].sum()*100:.1f}%)")
print(f"\nAverage inscription length: {object_dist['avg_length'].mean():.2f} signs")
print(f"Maximum inscription length: {object_dist['max_length'].max()} signs")
```
